```python
# Retail Sales Data
```

```python
sales_df = pd.read_csv("train.csv")
```

```python
display(sales_df.head())
```

```python
print(sales_df.info())
```

```python
print(sales_df.shape)
```

```python
print(sales_df.duplicated().sum())
```

```python
sales_df = sales_df.drop_duplicates()
```

```python
print(sales_df.duplicated().sum())
```

```python
print(sales_df.isna().sum())
```

```python
sales_df = sales_df.rename(columns={
    "PassengerId": "Passenger_ID",
    "Pclass": "Passenger_Class",
    "SibSp": "Sibling_Spouse",
    "Parch": "Parent_Child"
})
```

```python
sales_df["Name"] = (
    sales_df["Name"]
    .str.title()
    .str.strip()
)
```

```python
sales_df["Sex"] = (
    sales_df["Sex"]
    .str.title()
    .str.strip()
)
```

```python
sales_df["Embarked"] = (
    sales_df["Embarked"]
    .fillna("Unknown")
    .str.upper()
)
```

```python
sales_df["Age"] = sales_df["Age"].fillna(
    sales_df["Age"].median()
)
```

```python
sales_df["Fare"] = pd.to_numeric(
    sales_df["Fare"],
    errors="coerce"
)
```

```python
Q1 = sales_df["Fare"].quantile(0.25)
```

```python
Q3 = sales_df["Fare"].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR
```

```python
sales_df = sales_df[
    (sales_df["Fare"] >= lower_bound) &
    (sales_df["Fare"] <= upper_bound)
]
```

```python
display(sales_df.head())
```

```python
print(sales_df.dtypes)
```

```python
assert sales_df.duplicated().sum() == 0
```

```python
print("Validation Successful!")
```

```python
sales_df.to_csv("cleaned_retail_sales_dataset.csv", index=False)
```

```python
# Data Science Job Postings
```

```python
jobs_df = pd.read_csv("glassdoor job posting.csv")
```

```python
display(jobs_df.head())
```

```python
print(jobs_df.info())
```

```python
print(jobs_df.shape)
```

```python
print(jobs_df.duplicated().sum())
```

```python
jobs_df = jobs_df.drop_duplicates()
```

```python
print(jobs_df.duplicated().sum())
```

```python
print(jobs_df.isna().sum())
```

```python
jobs_df = jobs_df.rename(columns={
    "Job Title": "Job_Title",
    "Salary Estimate": "Salary_Estimate",
    "Job Description": "Job_Description",
    "Company Name": "Company_Name",
    "Type of ownership": "Ownership_Type"
})
```

```python
print(jobs_df.columns)
```

```python
jobs_df["Job_Title"] = (
    jobs_df["Job_Title"]
    .fillna("Unknown")
    .astype(str)
    .str.title()
    .str.strip()
)
```

```python
jobs_df["Company_Name"] = (
    jobs_df["Company_Name"]
    .fillna("Unknown")
    .astype(str)
    .str.replace(r"\n.*", "", regex=True)
    .str.strip()
)
```

```python
jobs_df["Location"] = (
    jobs_df["Location"]
    .fillna("Unknown")
    .astype(str)
    .str.title()
    .str.strip()
)
```

```python
jobs_df["Industry"] = (
    jobs_df["Industry"]
    .fillna("Unknown")
    .astype(str)
    .str.title()
    .str.strip()
)
```

```python
jobs_df["Sector"] = (
    jobs_df["Sector"]
    .fillna("Unknown")
    .astype(str)
    .str.title()
    .str.strip()
)
```

```python
jobs_df["Job_Description"] = (
    jobs_df["Job_Description"]
    .fillna("Not Available")
    .astype(str)
    .str.strip()
)
```

```python
jobs_df["Salary_Estimate"] = pd.to_numeric(
    jobs_df["Salary_Estimate"],
    errors="coerce"
)
```

```python
Q1 = jobs_df["Salary_Estimate"].quantile(0.25)

Q3 = jobs_df["Salary_Estimate"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR
```

```python
jobs_df = jobs_df[
    (jobs_df["Salary_Estimate"] >= lower_bound) &
    (jobs_df["Salary_Estimate"] <= upper_bound)
]
```

```python
display(jobs_df.head())
```

```python
print(jobs_df.dtypes)
```

```python
assert jobs_df.duplicated().sum() == 0
```

```python
print("Validation Successful!")
```

```python
jobs_df.to_csv(
    "cleaned_data_science_jobs_dataset.csv",
    index=False
)
```

```python
print("Cleaned dataset saved successfully!")
```

```python
import pandas as pd
```

```python
taxi_df = pd.read_csv("data.csv")
```

```python
taxi_df = pd.read_csv("data.csv")
```

```python
display(taxi_df.head())
```

```python
print(taxi_df.info())
```

```python
print(taxi_df.shape)
```

```python
print(taxi_df.duplicated().sum())
```

```python
taxi_df = taxi_df.drop_duplicates()
```

```python
print(taxi_df.duplicated().sum())
```

```python
print(taxi_df.isna().sum())
```

```python
print(taxi_df.columns)
```

```python
taxi_df = taxi_df.rename(columns={
    "passenger_count": "Passenger_Count",
    "trip_distance": "Trip_Distance",
    "fare_amount": "Fare_Amount",
    "pickup_longitude": "Pickup_Longitude",
    "pickup_latitude": "Pickup_Latitude",
    "dropoff_longitude": "Dropoff_Longitude",
    "dropoff_latitude": "Dropoff_Latitude"
})
```

```python
taxi_df["Passenger_Count"] = pd.to_numeric(
    taxi_df["Passenger_Count"],
    errors="coerce"
)
```

```python
taxi_df["Fare_Amount"] = pd.to_numeric(
    taxi_df["Fare_Amount"],
    errors="coerce"
)
```

```python
taxi_df["Trip_Distance"] = pd.to_numeric(
    taxi_df["Trip_Distance"],
    errors="coerce"
)
```

```python
taxi_df["Passenger_Count"] = taxi_df["Passenger_Count"].fillna(
    taxi_df["Passenger_Count"].median()
)
```

```python
taxi_df = taxi_df[
    (taxi_df["Pickup_Latitude"].between(-90, 90)) &
    (taxi_df["Dropoff_Latitude"].between(-90, 90)) &
    (taxi_df["Pickup_Longitude"].between(-180, 180)) &
    (taxi_df["Dropoff_Longitude"].between(-180, 180))
]
```

```python
Q1 = taxi_df["Fare_Amount"].quantile(0.25)
Q3 = taxi_df["Fare_Amount"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
taxi_df = taxi_df[
    (taxi_df["Fare_Amount"] >= lower_bound) &
    (taxi_df["Fare_Amount"] <= upper_bound)
]
```

```python
display(taxi_df.head())
```

```python
print(taxi_df.dtypes)
```

```python
assert taxi_df.duplicated().sum() == 0
```

```python
print("Validation Successful!")
```

```python
taxi_df.to_csv(
    "cleaned_nyc_taxi_dataset.csv",
    index=False
)
```
